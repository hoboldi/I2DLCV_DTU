```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
csv_ss = ""
csv_eb = ""

df_ss = pd.read_csv(csv_ss)
df_eb = pd.read_csv(csv_eb)
```

```python
iou_thresholds = [0.3, 0.5, 0.7]

def summarize_method(df, iou_thresholds):
    summary = {}
    for t in iou_thresholds:
        col_name = f"recall_{t:.2f}"
        if col_name not in df.columns:
            raise ValueError(f"Column {col_name} not found in CSV")
        summary[t] = df.groupby("num_proposals")[col_name].mean()
    # Average IoU
    summary["avg_iou"] = df.groupby("num_proposals")["avg_iou"].mean()
    return summary

summary_ss = summarize_method(df_ss, iou_thresholds)
summary_eb = summarize_method(df_eb, iou_thresholds)
```

```python
print("### Selective Search Summary")
pd.DataFrame(summary_ss)

print("\n### EdgeBoxes Summary")
pd.DataFrame(summary_eb)
```

```python
plt.figure(figsize=(10,6))
for t in iou_thresholds:
    plt.plot(summary_ss[t].index, summary_ss[t], label=f"SS recall@{t:.2f}")
    plt.plot(summary_eb[t].index, summary_eb[t], '--', label=f"EB recall@{t:.2f}")
plt.xlabel("Number of proposals")
plt.ylabel("Average Recall")
plt.title("Selective Search vs EdgeBoxes")
plt.legend()
plt.grid(True)
plt.show()
```
